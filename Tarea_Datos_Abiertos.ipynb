```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Define las columnas que representan los grupos de edad.
columnas_edad = ['Menor a 1', 'De 1 a 4', 'De 5 a 9', 'De 10 a 14', 'De 15 a 19', 'De 20 a 49', 'De 50 a 64', '65 Y MAS']

# Imprime un encabezado para la salida.
print("Enfermedad más común por grupo de edad:")

# Itera sobre cada columna de grupo de edad.
for columna in columnas_edad:
    # Filtra el DataFrame para las filas donde hay casos en el grupo de edad actual.
    df_filtrado = df[df[columna] > 0]

    # Encuentra el diagnóstico más común para el grupo de edad filtrado.
    diagnostico_principal = df_filtrado.groupby('diagnostico_ingreso')[columna].sum().idxmax()

    # Obtiene la cantidad de casos para ese diagnóstico principal.
    conteo_maximo = df_filtrado.groupby('diagnostico_ingreso')[columna].sum().max()

    # Imprime el resultado para el grupo de edad actual.
    print(f"  Grupo de Edad: {columna}: {diagnostico_principal} (Cantidad: {conteo_maximo})")
```
