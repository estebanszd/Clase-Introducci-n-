# **Nivel 2: Distribuciones y Cobertura Territorial**
## Objetivo: entender cómo se distribuye la red de acceso financiero a nivel departamental e identificar brechas territoriales.

```python
# Se limpia y convierte la columna de corresponsales activos a numérica, luego se agrupan por departamento.
df_ultimo['NRO CORRESPONSALES ACTIVOS'] = pd.to_numeric(
    df_ultimo['NRO CORRESPONSALES ACTIVOS'].astype(str).str.replace(',', ''), errors='coerce'
).fillna(0)

corresponsales_depto = (
    df_ultimo[df_ultimo['TIPO']=='CORRESPONSALES']
    .groupby('DEPARTAMENTO')['NRO CORRESPONSALES ACTIVOS']
    .sum()
    .sort_values()
)
```

```python
# Se genera un gráfico de barras horizontales para visualizar los corresponsales activos por departamento.
# El formateador mejora la legibilidad de los números grandes en el eje X.
fig, ax = plt.subplots(figsize=(10, 8))
ax.barh(corresponsales_depto.index, corresponsales_depto.values, edgecolor='white', linewidth=0.4)
ax.set_xlabel('Nro. Corresponsales Activos')
ax.set_title(f'Corresponsales Activos por Departamento — Corte {ultima_fecha.strftime("%d/%m/%Y")}')
ax.xaxis.set_major_formatter(mticker.FuncFormatter(lambda x,_: f'{int(x):,}'))
plt.tight_layout()
```

```python
# Se identifica y se muestra el número de municipios sin corresponsales activos, agrupados por departamento.
df_corr = df_ultimo[df_ultimo['TIPO']=='CORRESPONSALES']
cobertura_mun = df_corr.groupby(['DEPARTAMENTO','MUNICIPIO'])['NRO CORRESPONSALES ACTIVOS'].sum()
sin_cobertura = cobertura_mun[cobertura_mun == 0].reset_index()

print(f'Municipios con cero corresponsales activos: {len(sin_cobertura)}')
print()
print('Por departamento:')
print(sin_cobertura.groupby('DEPARTAMENTO').size().sort_values(ascending=False))
```

```python
# Se imprime el departamento con municipios sin cobertura, dado que solo hay uno.
sin_cob_dep = sin_cobertura.groupby('DEPARTAMENTO').size().sort_values(ascending=False).head(12)
sin_cob_dep
```

```python
# Se usa el método IQR para identificar valores atípicos (outliers) en los corresponsales activos.
col = 'NRO CORRESPONSALES ACTIVOS'
data_corr = df_corr[col]

Q1 = data_corr.quantile(0.25)
Q3 = data_corr.quantile(0.75)
IQR = Q3 - Q1

limite_superior = Q3 + 1.5 * IQR

print(f'Q1: {Q1:.0f} | Q3: {Q3:.0f} | IQR: {IQR:.0f}')
print(f'Límite superior (outlier): {limite_superior:.0f}')
```

```python
# Se filtran y muestran los registros que son outliers, aclarando que no son errores de datos.
outliers_corr = df_corr[df_corr[col] > limite_superior][
    ['NOMBRE DE LA  ENTIDAD','DEPARTAMENTO','MUNICIPIO', col]
]
print(f'Registros outlier: {len(outliers_corr):,}')
print()
print('Top 8 outliers:')
print(outliers_corr.sort_values(col, ascending=False).head(8).to_string(index=False))
print()
print('→ Los outliers no son errores: corresponden a entidades con redes masivas de corresponsales')
print('  tercerizados especializados en microfinanzas o pagos (Bancamía, Banco de Occidente).')
```

```python
# Se limpian y convierten a numérico las columnas 'NRO DEPOSITOS' y 'SALDO TOTAL CTA AHORROS'.
# Se generan histogramas para visualizar su distribución, excluyendo el 1% superior y usando un formato personalizado para ejes.
df_ultimo['NRO DEPOSITOS'] = pd.to_numeric(
    df_ultimo['NRO DEPOSITOS'].astype(str).str.replace(',', ''), errors='coerce'
).fillna(0)
df_ultimo['SALDO TOTAL CTA AHORROS'] = pd.to_numeric(
    df_ultimo['SALDO TOTAL CTA AHORROS'].astype(str).str.replace(',', ''), errors='coerce'
).fillna(0)

datos_dep    = df_ultimo[df_ultimo['NRO DEPOSITOS'] > 0]['NRO DEPOSITOS']
datos_ahorro = df_ultimo[df_ultimo['SALDO TOTAL CTA AHORROS'] > 0]['SALDO TOTAL CTA AHORROS']

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

q99_dep = datos_dep.quantile(0.99)
axes[0].hist(datos_dep[datos_dep <= q99_dep], bins=50, edgecolor='white', linewidth=0.3)
axes[0].set_xlabel('Número de Depósitos')
axes[0].set_ylabel('Frecuencia')
axes[0].set_title('Distribución: Nro. Depósitos (sin top 1%)')
axes[0].xaxis.set_major_formatter(plt.FuncFormatter(lambda x,_: f'{x:,.0f}'))

q99_aho = datos_ahorro.quantile(0.99)
axes[1].hist(datos_ahorro[datos_ahorro <= q99_aho], bins=50, edgecolor='white', linewidth=0.3)
axes[1].set_xlabel('Saldo Cuentas de Ahorro (COP)')
axes[1].set_ylabel('Frecuencia')
axes[1].set_title('Distribución: Saldo Ctas. Ahorro (sin top 1%)')

def format_magnitude(x, _):
    if x >= 1_000_000_000_000:
        return f'{x/1_000_000_000_000:,.0f}T'
    elif x >= 1_000_000_000:
        return f'{x/1_000_000_000:,.0f}B'
    elif x >= 1_000_000:
        return f'{x/1_000_000:,.0f}M'
    elif x >= 1_000:
        return f'{x/1_000:,.0f}K'
    else:
        return f'{x:,.0f}'

axes[1].xaxis.set_major_formatter(plt.FuncFormatter(format_magnitude))
axes[1].xaxis.set_major_locator(mticker.MaxNLocator(nbins=5))

plt.tight_layout()
plt.show()

print("")
print('Ambas distribuciones son sesgadas a la derecha (muchos valores bajos, pocos muy altos).')
print("")
print('Esto es típico de datos financieros territoriales con alta heterogeneidad entre municipios.')
```

```python
# Se limpian y convierten a numérico las columnas clave, luego se calcula la evolución trimestral de indicadores agregados.
for col in ['NRO CORRESPONSALES ACTIVOS', 'NRO DEPOSITOS', 'NRO TOTAL MICROCREDITO']:
    df_real[col] = pd.to_numeric(
        df_real[col].astype(str).str.replace(',', ''), errors='coerce'
    ).fillna(0)

serie_tiempo = (
    df_real
    .groupby('FECHA DE CORTE')
    .agg(
        NRO_CORR   = ('NRO CORRESPONSALES ACTIVOS', 'sum'),
        NRO_DEP    = ('NRO DEPOSITOS', 'sum'),
        NRO_MICRO  = ('NRO TOTAL MICROCREDITO', 'sum')
    )
    .reset_index()
    .sort_values('FECHA DE CORTE')
)

print('Evolución trimestral:')
print(serie_tiempo.to_string(index=False))
```

```python
# Se crea un gráfico de líneas con dos ejes Y para comparar la evolución de corresponsales, depósitos y microcréditos.
# Se observa el impacto de la pandemia de COVID-19.
fig, ax1 = plt.subplots(figsize=(11, 5))
ax2 = ax1.twinx()

l1, = ax1.plot(serie_tiempo['FECHA DE CORTE'], serie_tiempo['NRO_CORR'],
               color='red', marker='o', lw=2.5, label='Corresponsales (miles)')
l2, = ax2.plot(serie_tiempo['FECHA DE CORTE'], serie_tiempo['NRO_DEP'],
               color='blue', marker='s', lw=2.5, label='Nro. Depósitos (miles)')
l3, = ax2.plot(serie_tiempo['FECHA DE CORTE'], serie_tiempo['NRO_MICRO'],
               color='green', marker='^', lw=2, ls='--', label='Nro. Microcréditos (miles)')

ax1.set_ylabel('Corresponsales (miles)', color='black')
ax2.set_ylabel('Volumen Productos (miles)', color='black')
ax1.set_xlabel('Fecha de Corte')
ax1.set_title('Evolución Trimestral: Red de Corresponsales y Productos Financieros (2017–2021)')
ax1.tick_params(axis='x', rotation=30)
ax1.legend(handles=[l1, l2, l3], loc='upper left', fontsize=9)
plt.tight_layout()
plt.show()

print("")
print('Los microcréditos caen en 2020-Q2 (COVID-19) y se recuperan en 2020-Q3.')
print('Los corresponsales también cayeron pero se recuperaron más rápido.')
```

# **Nivel 4: Brecha de Género en Tenencia de Productos**
## Objetivo: calcular la proporción de tenencia por sexo en al menos tres productos, comparar por región y tipo de entidad, e interpretar en el contexto de política de inclusión financiera.


```python
# Se limpian y convierten a numérico las columnas relacionadas con el género para diversos productos.
# Se calculan las proporciones de tenencia por sexo y se resumen en `df_brechas`.
columnas_a_limpiar = [
    'NRO CTA AHORRO MUJERES', 'NRO CTA AHORRO HOMBRES',
    'NRO CREDITO CONSUMO MUJERES', 'NRO CREDITO CONSUMO HOMBRES',
    'NRO MICROCREDITO MUJERES', 'NRO MICROCREDITO HOMBRES',
    'NRO CREDITO VIVIENDA MUJERES', 'NRO CREDITO VIVIENDA HOMBRES'
]

for columna in columnas_a_limpiar:
    if columna in df_ultimo.columns:
        df_ultimo[columna] = pd.to_numeric(
            df_ultimo[columna].astype(str).str.replace(',', ''), errors='coerce'
        ).fillna(0)

productos = [
    (df_ultimo[df_ultimo['TIPO']=='CUENTAS DE AHORRO'],
     'NRO CTA AHORRO MUJERES','NRO CTA AHORRO HOMBRES','Cuentas de Ahorro'),
    (df_ultimo[df_ultimo['TIPO']=='CREDITO DE CONSUMO'],
     'NRO CREDITO CONSUMO MUJERES','NRO CREDITO CONSUMO HOMBRES','Crédito Consumo'),
    (df_ultimo[df_ultimo['TIPO']=='MICROCREDITO'],
     'NRO MICROCREDITO MUJERES','NRO MICROCREDITO HOMBRES','Microcrédito'),
    (df_ultimo[df_ultimo['TIPO']=='CREDITO DE VIVIENDA'],
     'NRO CREDITO VIVIENDA MUJERES','NRO CREDITO VIVIENDA HOMBRES','Crédito Vivienda'),
]

filas = []
for sub_df, col_mujeres, col_hombres, nombre in productos:
    conteo_mujeres = sub_df[col_mujeres].sum()
    conteo_hombres = sub_df[col_hombres].sum()
    total = conteo_mujeres + conteo_hombres
    filas.append({
        'Producto': nombre,
        'Mujeres': int(conteo_mujeres),
        'Hombres': int(conteo_hombres),
        '% Mujeres': round(conteo_mujeres/total*100, 1) if total > 0 else 0,
        '% Hombres': round(conteo_hombres/total*100, 1) if total > 0 else 0,
        'Ratio M/H': round(conteo_mujeres/conteo_hombres, 3) if conteo_hombres > 0 else None
    })

df_brechas = pd.DataFrame(filas)
df_brechas
```

```python
# Se visualizan las proporciones de tenencia por género para cada producto mediante un gráfico de barras agrupadas.
# Se incluye una línea de paridad y etiquetas de porcentaje para una mejor comprensión.
x = np.arange(len(df_brechas))
ancho = 0.35

fig, ax = plt.subplots(figsize=(10, 5))
barras_m = ax.bar(x - ancho/2, df_brechas['% Mujeres'], ancho, label='Mujeres', color='pink', edgecolor='white')
barras_h = ax.bar(x + ancho/2, df_brechas['% Hombres'], ancho, label='Hombres', color='skyblue', edgecolor='white')
ax.axhline(50, color='gray', ls='--', lw=1, alpha=0.7, label='Paridad (50%)')
ax.set_ylabel('Participación (%)')
ax.set_xticks(x)
ax.set_xticklabels(df_brechas['Producto'])
ax.set_title('Proporción de Tenencia por Género en Productos Financieros (2021-Q1)')
ax.legend()
ax.set_ylim(0, 75)
for barra in barras_m:
    ax.text(barra.get_x()+barra.get_width()/2, barra.get_height()+0.5, f'{barra.get_height():.1f}%', ha='center', fontsize=8, color='red')
for barra in barras_h:
    ax.text(barra.get_x()+barra.get_width()/2, barra.get_height()+0.5, f'{barra.get_height():.1f}%', ha='center', fontsize=8, color='blue')
plt.tight_layout()
plt.show()
```

```python
# Se filtra el DataFrame para 'CUENTAS DE AHORRO', se agrupan por departamento y se calcula la participación porcentual de mujeres.
df_ahorro = df_ultimo[df_ultimo['TIPO']=='CUENTAS DE AHORRO']
brecha_depto = df_ahorro.groupby('DEPARTAMENTO').agg(
    M = ('NRO CTA AHORRO MUJERES','sum'),
    H = ('NRO CTA AHORRO HOMBRES','sum')
).reset_index()

brecha_depto = brecha_depto[(brecha_depto['M']>0)|(brecha_depto['H']>0)].copy()
brecha_depto['PCT_M'] = brecha_depto['M'] / (brecha_depto['M']+brecha_depto['H']) * 100
brecha_depto = brecha_depto.sort_values('PCT_M', ascending=True)

print('Participación femenina en cuentas de ahorro por departamento:')
print(brecha_depto[['DEPARTAMENTO','PCT_M']].to_string(index=False))
```

```python
# Se calcula la participación porcentual de mujeres en cuentas de ahorro, desagregada por tipo de entidad.
brecha_entidad = df_ahorro.groupby('TIPO_ENT').agg(
    M = ('NRO CTA AHORRO MUJERES','sum'),
    H = ('NRO CTA AHORRO HOMBRES','sum')
).reset_index()
brecha_entidad = brecha_entidad[(brecha_entidad['M']>0)|(brecha_entidad['H']>0)].copy()
brecha_entidad['PCT_M'] = brecha_entidad['M'] / (brecha_entidad['M']+brecha_entidad['H']) * 100

print('Participación femenina por tipo de entidad:')
print(brecha_entidad[['TIPO_ENT','PCT_M']].to_string(index=False))
```

# **Nivel 5: Componente de Criterio Libre**
## Pregunta analítica propia

**¿Cómo ha evolucionado la participación de las mujeres en microcrédito entre 2017 y 2021, y difiere esta tendencia según el tipo de entidad financiera?**

**Justificación:** El microcrédito es el producto con mayor participación femenina en el sistema. Colombia lo priorizó como herramienta de inclusión desde 2010. Sin embargo, la pandemia de COVID-19 en 2020 pudo haber revertido estos avances de forma diferenciada por tipo de entidad. Analizar la evolución permite verificar si el avance de género fue sostenido o interrumpido, y quién respondió mejor.


```python
# Se filtra el DataFrame para microcrédito, se limpian y convierten a numérico las columnas de mujeres y hombres.
# Se agrupan los datos por fecha para calcular la participación porcentual de mujeres en microcréditos a lo largo del tiempo.
df_micro = df_mun[df_mun['TIPO']=='MICROCREDITO'].copy()

for col in ['NRO MICROCREDITO MUJERES', 'NRO MICROCREDITO HOMBRES']:
    df_micro[col] = pd.to_numeric(
        df_micro[col].astype(str).str.replace(',', ''), errors='coerce'
    ).fillna(0).astype(float)

serie_genero = df_micro.groupby('FECHA DE CORTE').agg(
    M = ('NRO MICROCREDITO MUJERES','sum'),
    H = ('NRO MICROCREDITO HOMBRES','sum')
).reset_index()

# Se maneja la división por cero para evitar errores.
serie_genero['PCT_M'] = serie_genero.apply(
    lambda row: (row['M'] / (row['M'] + row['H'])) * 100 if (row['M'] + row['H']) > 0 else 0,
    axis=1
)
serie_genero = serie_genero.sort_values('FECHA DE CORTE')

print('Evolución trimestral % mujeres en microcrédito:')
print(serie_genero[['FECHA DE CORTE','PCT_M']].to_string(index=False))
```

```python
# Se calcula la participación femenina en microcrédito desagregada por fecha de corte y tipo de entidad.
# Se reemplazan las divisiones por cero con NaN para un cálculo robusto.
serie_por_entidad = df_micro.groupby(['FECHA DE CORTE','TIPO_ENT']).agg(
    M = ('NRO MICROCREDITO MUJERES','sum'),
    H = ('NRO MICROCREDITO HOMBRES','sum')
).reset_index()

serie_por_entidad['PCT_M'] = (
    serie_por_entidad['M'] /
    (serie_por_entidad['M'] + serie_por_entidad['H']).replace(0, np.nan) * 100
)
serie_por_entidad = serie_por_entidad.sort_values('FECHA DE CORTE')
serie_por_entidad
```

```python
# Se crean dos gráficos para visualizar la evolución de la participación femenina en microcrédito: global y por tipo de entidad.
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

#Gráfico izquierdo: tendencia global
ax = axes[0]
ax.plot(serie_genero['FECHA DE CORTE'], serie_genero['PCT_M'],
        color='#cb4335', marker='o', lw=2.5, label='% Mujeres')
ax.axhline(50, color='gray', ls='--', lw=1.5, alpha=0.7, label='Paridad (50%)')
ax.fill_between(serie_genero['FECHA DE CORTE'], 50, serie_genero['PCT_M'],
                alpha=0.15, where=serie_genero['PCT_M']>50, color='#cb4335')
# Se marca el inicio del COVID.
ax.axvline(pd.Timestamp('2020-03-31'), color='#d4ac0d', ls=':', lw=2, label='Inicio COVID (2020-Q1)')
ax.set_ylabel('% Titulares Mujeres')
ax.set_xlabel('Fecha de Corte')
ax.set_title('Evolución de la Participación Femenina en Microcrédito (2017–2021)')
ax.set_ylim(45, 70)
ax.tick_params(axis='x', rotation=30)
ax.legend(fontsize=9)
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x,_: f'{x:.0f}%'))

#Gráfico derecho: por tipo de entidad
paleta = {'Banco':'red','Sedpe':'green','Cooperativa':'blue','Otro':'purple'}
ax2 = axes[1]
for ent, grupo in serie_por_entidad.groupby('TIPO_ENT'):
    grupo = grupo.sort_values('FECHA DE CORTE')
    if grupo['PCT_M'].notna().sum() >= 3:
        ax2.plot(grupo['FECHA DE CORTE'], grupo['PCT_M'], marker='o', lw=2,
                label=ent, color=paleta.get(ent, '#1a5276'))
ax2.axhline(50, color='gray', ls='--', lw=1.5, alpha=0.7)
ax2.set_ylabel('% Titulares Mujeres')
ax2.set_xlabel('Fecha de Corte')
ax2.set_title('Participación Femenina en Microcrédito por Tipo de Entidad')
ax2.set_ylim(0, 80)
ax2.tick_params(axis='x', rotation=30)
ax2.legend(fontsize=9)
ax2.yaxis.set_major_formatter(plt.FuncFormatter(lambda x,_: f'{x:.0f}%'))

plt.tight_layout()
plt.show()
```

### Conclusiones del Componente Libre

1. **Tendencia de largo plazo:** la participación femenina en microcrédito pasó de 53.4% en 2017-Q4 a 58.8% en 2021-Q1 (+5.4 pp en tres años).

2. **Efecto COVID-19:** en 2020-Q2 se observa la única caída por debajo del 50% (49.9%), posiblemente porque el confinamiento afectó más a las mujeres con negocios informales en sectores como comercio y restaurantes.

3. **Rebote post-COVID:** en 2020-Q3 la participación femenina sube a 59.0%, su máximo histórico. El microcrédito fue el vehículo principal de recuperación económica para mujeres microempresarias.

4. **Diferencias por entidad:** los bancos muestran más varianza entre trimestres. Las entidades especializadas mantienen participación femenina más estable, coherente con su misión social.

5. **Implicación de política:** el patrón justifica la continuidad y refuerzo de programas de microcrédito dirigidos a mujeres como instrumento anticíclico. La rápida recuperación post-COVID valida la resiliencia de las microempresarias colombianas cuando tienen acceso a financiamiento formal.
