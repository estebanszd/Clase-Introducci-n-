#**Nivel 2: Limpieza**
##Objetivo: transformar cada columna al estado que permite análisis confiable.

```python
# Contamos los valores nulos en la columna 'fecha_prescripcion' antes de la conversión.
nulos_fecha = df['fecha_prescripcion'].isna().sum()

print(nulos_fecha)
```

```python
# Convertimos la columna 'fecha_prescripcion' a formato datetime, manejando diferentes formatos y errores.
df['fecha_prescripcion'] = pd.to_datetime(
    df['fecha_prescripcion'],
    dayfirst=True,
    format='mixed',
    errors='coerce'
)
```

```python
# Verificamos nuevamente la cantidad de nulos en 'fecha_prescripcion' después de la conversión.
nulos_fecha = df['fecha_prescripcion'].isna().sum()

print(nulos_fecha)
```

```python
# Calculamos la mediana de la columna 'fecha_prescripcion' para rellenar los valores nulos.
mediana_fecha = df['fecha_prescripcion'].dropna().sort_values().iloc[len(df['fecha_prescripcion'].dropna()) // 2]

# Rellenamos los NaT (Not a Time) con la mediana calculada.
df['fecha_prescripcion'] = df['fecha_prescripcion'].fillna(mediana_fecha)
```

```python
# Verificamos que no queden valores nulos en la columna 'fecha_prescripcion' después de rellenar.
nulos_fecha = df['fecha_prescripcion'].isna().sum()

print(nulos_fecha)
```

```python
# Mostramos las primeras filas del DataFrame para ver los cambios realizados en las fechas.
df.head()
```

```python
# Normalizamos los campos 'medicamento' y 'via_administracion': eliminamos espacios y convertimos a minúsculas.
df['medicamento'] = df['medicamento'].str.strip().str.lower()
df['via_administracion'] = df['via_administracion'].str.strip().str.lower()
```

```python
# Mostramos las primeras filas del DataFrame para verificar la normalización de texto.
df.head()
```

```python
# Creamos diccionarios de mapeo para unificar variantes de 'vias de administración' y 'medicamentos'.
mapeo_vias = {
    'via oral': 'oral',
    'vo': 'oral',
    'iv': 'intravenosa',
    'subcutanea': 'subcutanea',
    'sc': 'subcutanea',
    'oral': 'oral',
    'intravenosa': 'intravenosa'
}
mapeo_medicamentos = {
    'asa': 'aspirina',
    'aspirina': 'aspirina',
    'metformina 500': 'metformina',
    'metformina': 'metformina',
    'ibuprofeno': 'ibuprofeno',
    'enalapril': 'enalapril',
    'losartan': 'losartan'
}
```

```python
# Aplicamos los mapeos para estandarizar los valores en las columnas 'via_administracion' y 'medicamento'.
df['via_administracion'] = df['via_administracion'].replace(mapeo_vias)
df['medicamento'] = df['medicamento'].replace(mapeo_medicamentos)
```

```python
# Imprimimos los valores únicos de 'via_administracion' y 'medicamento' para verificar la unificación.
print(df['via_administracion'].unique())

print(df['medicamento'].unique())
```

```python
# Mostramos las primeras 20 filas del DataFrame para observar los cambios en las columnas unificadas.
df.head(20)
```

```python
# Imprimimos la cantidad de variantes únicas en la columna 'eps' antes de la unificación y su recuento.
print('Variantes únicas de EPS antes de unificar:', df['eps'].nunique())
print(df['eps'].value_counts())
```

```python
# Normalizamos la columna 'eps' convirtiendo a minúsculas y eliminando espacios.
df['eps'] = df['eps'].str.strip().str.lower()

# Creamos un diccionario de mapeo para unificar las diferentes representaciones de las EPS.
mapeo_eps = {
    'sanitas': 'eps sanitas',
    'eps sanitas': 'eps sanitas',
    'nueva eps': 'nueva eps',
    'nuevaeps': 'nueva eps',
    'sura': 'eps sura',
    'eps sura': 'eps sura',
    'compensar': 'compensar eps',
    'eps compensar': 'compensar eps',
    'salud total': 'salud total'
}

# Aplicamos el mapeo a la columna 'eps'.
df['eps'] = df['eps'].replace(mapeo_eps)
```

```python
# Imprimimos la cantidad de variantes únicas de 'eps' después de la unificación y su recuento.
print('Variantes únicas de eps despues de unificar:', df['eps'].nunique())
print(df['eps'].value_counts())
```

```python
# Mostramos las primeras 20 filas del DataFrame para verificar los cambios en la columna 'eps'.
df.head(20)
```

```python
# Mostramos estadísticas descriptivas para la columna 'peso_paciente_kg'.
print(df['peso_paciente_kg'].describe())
```

```python
# Calculamos el primer cuartil (Q1), tercer cuartil (Q3) y el rango intercuartílico (IQR) para 'peso_paciente_kg'.
Q1 = df['peso_paciente_kg'].quantile(0.25)
Q3 = df['peso_paciente_kg'].quantile(0.75)
IQR = Q3 - Q1

# Calculamos los límites inferior y superior para identificar outliers utilizando el método IQR.
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

# Imprimimos los valores calculados para análisis.
print(f'Q1: {Q1:.2f} | Q3: {Q3:.2f} | IQR: {IQR:.2f}')
print(f'Límite inferior: {limite_inferior:.2f} | Límite superior: {limite_superior:.2f}')
```

```python
# Identificamos los outliers en 'peso_paciente_kg' basándonos en los límites calculados.
outliers = df[(df['peso_paciente_kg'] < limite_inferior) | (df['peso_paciente_kg'] > limite_superior)]

# Imprimimos la cantidad de outliers detectados.
print(f'\nOutliers detectados: {len(outliers)}')
```

```python
# Calculamos la mediana de 'peso_paciente_kg' para rellenar los outliers y valores NaN.
mediana_peso = df['peso_paciente_kg'].median()

# Reemplazamos los outliers por la mediana.
df.loc[
    (df['peso_paciente_kg'] < limite_inferior) | (df['peso_paciente_kg'] > limite_superior),
    'peso_paciente_kg'
] = mediana_peso

# También rellenamos los NaN con la mediana.
df['peso_paciente_kg'] = df['peso_paciente_kg'].fillna(mediana_peso)

# Mostramos las estadísticas descriptivas de 'peso_paciente_kg' después del rellenado.
print(df['peso_paciente_kg'].describe())
```

```python
# Mostramos estadísticas descriptivas para la columna 'edad'.
print(df['edad'].describe())
```

```python
# Reemplazamos los valores de 'edad' menores a 0 por NaN, ya que no existe nadie menor de 0 años.
df.loc[df['edad'] < 0, 'edad'] = np.nan
```

```python
# Calculamos Q1, Q3 e IQR para la columna 'edad'.
Q1_edad = df['edad'].quantile(0.25)
Q3_edad = df['edad'].quantile(0.75)
IQR_edad = Q3_edad - Q1_edad

# Calculamos los límites inferior y superior para detectar outliers en 'edad', asegurando que el límite inferior no sea negativo.
limite_inf_edad = max(0, Q1_edad - 1.5 * IQR_edad)
limite_sup_edad = Q3_edad + 1.5 * IQR_edad

# Imprimimos los valores calculados.
print(f'Q1: {Q1_edad} | Q3: {Q3_edad} | IQR: {IQR_edad}')
print(f'Límite inferior: {limite_inf_edad:.2f} | Límite superior: {limite_sup_edad:.2f}')

# Identificamos los outliers en 'edad'.
outliers_edad = df[(df['edad'] < limite_inf_edad) | (df['edad'] > limite_sup_edad)]
print(f'\nOutliers de edad detectados: {len(outliers_edad)}')
print('Valores:', sorted(outliers_edad['edad'].dropna().unique()))
```

```python
# Calculamos la mediana de 'edad' para rellenar los outliers y valores NaN.
mediana_edad = df['edad'].median()

# Remplazamos los outliers en 'edad' por la mediana.
df.loc[
    (df['edad'] < limite_inf_edad) | (df['edad'] > limite_sup_edad),
    'edad'
] = mediana_edad
# Rellenamos cualquier valor NaN restante en 'edad' con la mediana.
df['edad'] = df['edad'].fillna(mediana_edad)
#Volvemos a edad un entero para que no queden decimales en edad
df['edad'] = df['edad'].astype(int)
```

```python
# Usamos expresiones regulares para extraer solo los números (incluyendo decimales)
df['dosis_mg'] = df['dosis_mg'].astype(str).str.extract(r'(\d+\.?\d*)').astype(float)

# Convertimos negativos a NaN para que sean imputados en el siguiente paso
df.loc[df['dosis_mg'] < 0, 'dosis_mg'] = np.nan

# Convertimos valores mayores de 2000 ya que no hay paciente que reciba más de 1.000 a excepción de unos outliers que aparecián con 50.000
df.loc[df['dosis_mg'] > 2000, 'dosis_mg'] = np.nan

# Usamos transform para mantener el tamaño original del dataframe
df['dosis_mg'] = df['dosis_mg'].fillna(
    df.groupby('medicamento')['dosis_mg'].transform('median')
)
```

#**Nivel 3: Transformación y validación**
##Objetivo: generar variables derivadas y garantizar la integridad del dataset final.

```python
# Definimos una función para clasificar la edad en rangos etarios.
def clasificar_edad(edad):
    if pd.isna(edad):
        return 'desconocido'
    elif edad < 18:
        return 'pediátrico'
    elif edad <= 64:
        return 'adulto'
    else:
        return 'adulto mayor'
```

```python
# Aplicamos la función 'clasificar_edad' para crear una nueva columna 'rango_etario'.
df['rango_etario'] = df['edad'].apply(clasificar_edad)

# Imprimimos la distribución de los rangos etarios.
print('Distribución de rango_etario:')
print(df['rango_etario'].value_counts())
```

```python
# Aplicamos la función 'clasificar_edad' para crear una nueva columna 'rango_etario'.
df['rango_etario'] = df['edad'].apply(clasificar_edad)

# Imprimimos la distribución de los rangos etarios.
print('Distribución de rango_etario:')
print(df['rango_etario'].value_counts())
```

```python
# Creamos una tabla resumen agrupando por EPS y Vía de Administración, calculando estadísticas de 'dosis_mg'.
tabla_resumen = df.groupby(['eps', 'via_administracion'])['dosis_mg'].agg(
    conteo_dosis='count',
    media_dosis='mean',
    mediana_dosis='median',
    desviacion_std_dosis='std',
    minimo_dosis='min',
    maximo_dosis='max'
).round(2).reset_index()
#Imprimoos tabla de resumen
print('Tabla resumen — Prescripciones por EPS y Vía de Administración:')
tabla_resumen
```
