```python
df = datos_crudos.copy()
```

```python
print(df.shape)
print(110* "*")
print(df.dtypes)
print(110* "*")
print(df.isnull().sum())
print(110* "*")
print(df.duplicated().sum())
print(110* "*")
df.head()
```

```python
#Eliminación de duplicados+
print(f"filas antes: {df.shape[0]}")
df = df.drop_duplicates()
print(f"Filas después: {df.shape[0]}")
```

```python
datos_crudos.columns
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.encode("ascii", errors="ignores")
    .str.decode("ascii")
)
print(df.columns.tolist())
```

```python
df.head()
```

```python
var1 = "EDaD "
```

```python
var1.strip().lower()
```

```python
#Nombre: capitalización consistente
df["nombre"] = df["nombre"].str.strip().str.title()
#Diagnóstico: Mayúsculas sin espacios
df["diagnostico"] = df["diagnostico"].str.strip().str.upper()
#Sexo: unificar categorías antes de convertir
mapa_sexo = {
    "m": "M", "masculino": "M",
    "f": "F", "femenino": "F",
    "999": np.nan
}
df["sexo"] = df["sexo"].str.strip().str.lower().map(mapa_sexo)
print(df["sexo"].value_counts(dropna=False))
```

```python
#peso_kg: reemplazar "N/A" string antes de convertir
df['peso_kg'] = df['peso_kg'].replace('N/A', np.nan)
df['peso_kg'] = pd.to_numeric(df['peso_kg'], errors='coerce')

#costo_usd: eeliminar $ y comas
df['costo_usd'] = (
    df['costo_usd']
    .str.replace('$', '', regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
    )
df['costo_usd'] = pd.to_numeric(df['costo_usd'], errors='coerce')
print(df[["peso_kg", "costo_usd"]].dtypes)
```

```python
df["fecha_ingreso"] = pd.to_datetime(df["fecha_ingreso"], dayfirst=False, errors="coerce")
df["fecha_alta"] = pd.to_datetime(df["fecha_alta"], dayfirst=False, errors="coerce")

```

```python
# Limpiar y convertir 'edad': descartar valores fuera de 0-105, mantener NaN y convertir a entero (nullable)
df['edad'] = df['edad'].apply(lambda x: x if 0 <= x <= 105 else np.nan)
# Calcular el promedio de edad antes de llenar los NaN
mean_edad = df['edad'].mean()
# Llenar los NaN con el promedio de edad (redondeado para la conversión a Int64Dtype)
df['edad'] = df['edad'].fillna(round(mean_edad))
df['edad'] = df['edad'].astype(pd.Int64Dtype())

# Limpiar y convertir 'n_medicamentos': rellenar NaN con 0 y convertir a entero
df['n_medicamentos'] = df['n_medicamentos'].fillna(0).astype(int)
```
