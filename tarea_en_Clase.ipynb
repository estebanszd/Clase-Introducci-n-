```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
movies = pd.read_csv("https://raw.githubusercontent.com/Gustavo-Bruges/movies/refs/heads/main/movies.csv")

ratings = pd.read_csv("https://raw.githubusercontent.com/Gustavo-Bruges/movies/refs/heads/main/ratings.csv")
```

```python
print(movies.shape)

print(ratings.shape)
```

```python
print("Movies dataset")
print(movies.head())

print("Ratings dataset")
print(ratings.head())
```

```python
movie_ratings = movies.merge(ratings, on="id")

movie_ratings.head()

movie_ratings.drop(["Unnamed: 0_x", "Unnamed: 0_y"], axis = 1, inplace = True )
```

```python

movie_ratings.head()
```

```python
movie_genres = pd.read_csv("https://raw.githubusercontent.com/Gustavo-Bruges/movies/refs/heads/main/movie_to_genres.csv")
```

```python
movie_genres.shape
```

```python
movie_genres.head()
```

```python
movie_genres.drop(["Unnamed: 0" ], axis = 1, inplace = True )
```

```python
movies.drop(["Unnamed: 0" ], axis = 1, inplace = True )
print()
```

```python
movie_genres.head()
```

```python
movie_generos = movies.merge(movie_genres, left_on="id", right_on="movie_id")

movie_generos.head()
```

```python
movie_generos.sort_values(by='popularity', ascending=False).head()
```

```python
movie_generos_action = movie_generos[movie_generos["genre"] == "Action"]

movie_generos_action
```

```python
movie_generos_action.shape
```

```python
movie_generos_sci = movie_generos[movie_generos["genre"] == "Science Fiction"]

movie_generos_sci
```

```python
movie_generos_sci.shape
```

```python
print(movie_generos_sci)

print(movie_generos_action)
```

```python
df_left = movie_generos_action.merge(movie_generos_sci, on="movie_id", how ="left", suffixes=("_action", "_sci"))
```

```python
df_left.shape
```

```python
df_left.head()
```

```python
df_right = movie_generos_action.merge(movie_generos_sci, on="movie_id", how ="right", suffixes=("_action", "_sci"))
```
